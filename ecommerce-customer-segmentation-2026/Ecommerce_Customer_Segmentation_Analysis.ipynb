### Numeric Data Validation

Key numeric fields were checked for impossible or invalid values.

- No customers were below age 18.
- No negative values were found in purchases, spending, average order value, CLV, acquisition cost, or days since last purchase.
- Some customers have negative profitability, which is retained because it may represent genuinely unprofitable customers rather than data errors.

```python
numeric_cols = [
    'total_purchases',
    'avg_order_value_usd',
    'total_spent_usd',
    'customer_lifetime_value_usd',
    'customer_acquisition_cost_usd',
    'customer_profitability_usd'
]

for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    print(f"{col}: {outliers} potential outliers")
```

## Business Question 1: Which customer segments generate the most revenue?

```python
segment_revenue = (
    df.groupby('customer_segment')['total_spent_usd']
      .sum()
      .sort_values(ascending=False)
)

segment_revenue
```

```python
segment_avg_spend = (
    df.groupby('customer_segment')['total_spent_usd']
      .mean()
      .sort_values(ascending=False)
)

segment_avg_spend
```

### Business Insight

The Consumer segment generates the highest total revenue, contributing approximately $1.14 billion.

However, average spending per customer is very similar across all customer segments, with Premium customers having the highest average spend at roughly $50.46K.

This suggests that the Consumer segment leads in total revenue primarily because it has the largest customer base, rather than because individual Consumer customers spend significantly more.

### Business Question 2: Which customers have the highest lifetime value?

```python
df.groupby('customer_segment')['customer_lifetime_value_usd'] \
  .mean() \
  .sort_values(ascending=False)
```

```python
df.nlargest(10, 'customer_lifetime_value_usd')[
    ['customer_id', 'customer_segment', 'customer_lifetime_value_usd',
     'total_spent_usd', 'loyalty_tier', 'rfm_category']
]
```

### Business Insight

Premium customers have the highest average customer lifetime value at approximately $58K, although the difference between segments is relatively small.

The highest-value individual customers have CLV values above $275K. Most of these customers belong to the Diamond loyalty tier and are classified as Champions or Loyal customers.

This suggests that loyalty status and RFM behavior may be stronger indicators of long-term customer value than customer segment alone.

### Business Question 3: Which high-value customers are at risk of churning?

```python
high_value_at_risk = df[
    (df['customer_value_category'].isin(['High', 'Very High'])) &
    (df['churn_risk_category'].isin(['High', 'Very High']))
]

high_value_at_risk.shape
```

```python
high_value_at_risk[
    [
        'customer_id',
        'customer_segment',
        'customer_value_category',
        'churn_risk_category',
        'customer_lifetime_value_usd',
        'total_spent_usd',
        'activity_status',
        'rfm_category'
    ]
].head(20)
```

```python
high_value_at_risk['customer_lifetime_value_usd'].sum()
```

### Business Insight

A total of 1,511 high-value customers are classified as having High or Very High churn risk.

Together, these customers represent approximately $128.57 million in customer lifetime value.

This group should be prioritized for retention efforts because losing these customers could have a significant impact on long-term business value. Retention campaigns, personalized offers, and proactive engagement should focus on high-value customers showing elevated churn risk.

## Business Question 4: How does purchase frequency affect customer value?

```python
df.groupby('purchase_frequency')[
    ['total_spent_usd',
     'customer_lifetime_value_usd',
     'customer_profitability_usd']
].mean().sort_values(
    by='customer_lifetime_value_usd',
    ascending=False
)
```

```python
df.groupby('purchase_frequency')['customer_id'].count().sort_values(ascending=False)
```

### Business Insight

Customers who purchase Weekly have the highest average customer lifetime value and profitability.

Rarely purchasing customers have the lowest average customer lifetime value and profitability.

Because the number of customers is relatively balanced across purchase-frequency groups, the results suggest that more frequent purchasing behavior is generally associated with higher long-term customer value.

The business could focus on encouraging lower-frequency customers to purchase more often through loyalty rewards, targeted promotions, and personalized offers.

### Business Question 5: Which loyalty tiers generate the most revenue?

```python
loyalty_revenue = (
    df.groupby('loyalty_tier')['total_spent_usd']
      .sum()
      .sort_values(ascending=False)
)

loyalty_revenue
```

```python
loyalty_avg_spend = (
    df.groupby('loyalty_tier')['total_spent_usd']
      .mean()
      .sort_values(ascending=False)
)

loyalty_avg_spend
```

```python
df['loyalty_tier'].value_counts()
```

### Business Insight

Diamond customers generate the most revenue, contributing approximately $1.92 billion, or about 76% of total revenue.

They also have the highest average spend at approximately $94.7K per customer.

Spending declines substantially across lower loyalty tiers, while customers with no loyalty tier have the lowest average spend.

This suggests that loyalty tier is strongly associated with customer value. The business should prioritize retaining Diamond and Platinum customers while also creating strategies to move lower-tier customers upward through the loyalty program.

### Business Question 6: Which customer groups have the highest profitability?

```python
segment_profitability = (
    df.groupby('customer_segment')['customer_profitability_usd']
      .mean()
      .sort_values(ascending=False)
)

segment_profitability
```

```python
rfm_profitability = (
    df.groupby('rfm_category')['customer_profitability_usd']
      .mean()
      .sort_values(ascending=False)
)

rfm_profitability
```

```python
df.nlargest(10, 'customer_profitability_usd')[
    [
        'customer_id',
        'customer_segment',
        'rfm_category',
        'loyalty_tier',
        'customer_profitability_usd'
    ]
]
```

### Business Insight

Average profitability is relatively similar across the four customer segments, with Premium customers having the highest average profitability at approximately $57.9K.

However, profitability differs substantially across RFM categories. Champions generate the highest average profitability at approximately $87.4K, compared with only about $3.0K for customers classified as Need Attention.

The most profitable individual customers are also all Diamond loyalty members and are primarily classified as Champions.

This suggests that RFM behavior and loyalty status are stronger indicators of customer profitability than broad customer segment alone.

### Business Question 7: How does customer engagement relate to spending?

```python
engagement_spending = (
    df.groupby('engagement_level')[
        ['total_spent_usd',
         'customer_lifetime_value_usd',
         'customer_profitability_usd']
    ]
    .mean()
    .sort_values(
        by='total_spent_usd',
        ascending=False
    )
)

engagement_spending
```

```python
df['engagement_level'].value_counts()
```

### Business Insight

Customer engagement level shows little difference in average spending, customer lifetime value, or profitability.

Low-engagement customers have the highest average spending at approximately $50.4K, but the difference compared with Medium- and High-engagement customers is very small.

This suggests that engagement level alone is not a strong indicator of customer financial value in this dataset. Other factors, such as loyalty tier and RFM category, appear to be more useful for identifying valuable customers.

### Business Question 10: What customer groups should the business prioritize for retention?

```python
retention_priority = (
    high_value_at_risk
    .groupby('customer_segment')
    .agg(
        customers=('customer_id', 'count'),
        total_clv=('customer_lifetime_value_usd', 'sum'),
        avg_clv=('customer_lifetime_value_usd', 'mean'),
        avg_churn_risk=('churn_risk_score', 'mean')
    )
    .sort_values(by='total_clv', ascending=False)
)

retention_priority
```

```python
high_value_at_risk['rfm_category'].value_counts()
```

```python
high_value_at_risk['activity_status'].value_counts()
```

### Business Insight

The business should prioritize 1,511 high-value customers who are currently classified as having High or Very High churn risk. Together, these customers represent approximately $128.6 million in customer lifetime value.

The Consumer segment represents the largest retention opportunity, with 674 customers and approximately $57.0 million in CLV at risk.

Approximately 80% of these priority customers are already inactive, while most are classified as Loyal or Potential Loyalists based on RFM behavior.

The business should prioritize targeted re-engagement campaigns for inactive high-value customers, especially Consumers and customers with Loyal or Potential Loyalist status. Personalized offers, loyalty incentives, and win-back campaigns could help protect significant long-term customer value.

## Visual Exploratory Analysis

The following visualizations summarize the key findings from the exploratory analysis and highlight customer revenue, value, loyalty, profitability, and retention opportunities.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
segment_revenue_chart = (
    df.groupby('customer_segment')['total_spent_usd']
      .sum()
      .sort_values(ascending=False)
      / 1_000_000_000
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=segment_revenue_chart.index,
    y=segment_revenue_chart.values
)

plt.title('Total Revenue by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Revenue (USD Billions)')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### Key Finding

The Consumer segment generates the highest total revenue. However, earlier analysis showed that average spending per customer is similar across segments, meaning Consumer leads primarily because it has the largest customer base.

```python
loyalty_clv_chart = (
    df.groupby('loyalty_tier')['customer_lifetime_value_usd']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=loyalty_clv_chart.index,
    y=loyalty_clv_chart.values
)

plt.title('Average Customer Lifetime Value by Loyalty Tier')
plt.xlabel('Loyalty Tier')
plt.ylabel('Average CLV (USD)')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### Key Finding

Higher loyalty tiers are associated with substantially greater customer lifetime value. Diamond customers have the highest average CLV, while customers with no loyalty tier have the lowest.

```python
clv_at_risk_chart = (
    high_value_at_risk
    .groupby('customer_segment')['customer_lifetime_value_usd']
    .sum()
    .sort_values(ascending=False)
    / 1_000_000
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=clv_at_risk_chart.index,
    y=clv_at_risk_chart.values
)

plt.title('Customer Lifetime Value at Risk by Segment')
plt.xlabel('Customer Segment')
plt.ylabel('CLV at Risk (USD Millions)')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### Key Finding

The Consumer segment represents the largest retention opportunity, with approximately $57 million in customer lifetime value at risk.

High-value customers with elevated churn risk should be prioritized for targeted retention and re-engagement campaigns.

```python
rfm_profit_chart = (
    df.groupby('rfm_category')['customer_profitability_usd']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=rfm_profit_chart.index,
    y=rfm_profit_chart.values
)

plt.title('Average Customer Profitability by RFM Category')
plt.xlabel('RFM Category')
plt.ylabel('Average Profitability (USD)')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### Key Finding

Champions generate the highest average profitability by a wide margin, followed by Loyal customers.

Customers classified as At Risk or Need Attention generate substantially lower average profitability, reinforcing the importance of retaining strong RFM customers and re-engaging valuable customers before they decline further.

```python
activity_risk_chart = (
    high_value_at_risk['activity_status']
    .value_counts()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=activity_risk_chart.index,
    y=activity_risk_chart.values
)

plt.title('High-Value Customers at Risk by Activity Status')
plt.xlabel('Activity Status')
plt.ylabel('Number of Customers')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### Key Finding

Most high-value customers with elevated churn risk are already Inactive.

This indicates that retention efforts should focus heavily on win-back and re-engagement campaigns for valuable customers who have stopped purchasing.
